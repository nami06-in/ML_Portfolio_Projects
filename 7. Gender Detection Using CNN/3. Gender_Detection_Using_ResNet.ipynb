```python
data = np.load("/content/drive/MyDrive/Colab Notebooks/gender_dataset_ResNet.npz")

X = data["X"]
y = data["y"]
```

```python
X.shape
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.15,
    stratify=y,
    random_state=42
)

x_train_final, x_val, y_train_final, y_val = train_test_split(
    x_train,
    y_train,
    test_size=0.1765,
    stratify=y_train,
    random_state=42
)
```

```python
print("Train:", np.bincount(y_train_final))
print("Validation:", np.bincount(y_val))
print("Test:", np.bincount(y_test))
```

class distribution remains consistent across splits

```python
train_datagen = ImageDataGenerator(
    rotation_range=15,
    width_shift_range=0.08,
    height_shift_range=0.08,
    zoom_range=0.15,
    horizontal_flip=True,
    brightness_range=[0.8,1.2],
    fill_mode="nearest"
)

train_generator = train_datagen.flow(
    x_train_final,
    y_train_final,
    batch_size=32,
    shuffle=True
)
```

```python
val_datagen = ImageDataGenerator()

val_generator = val_datagen.flow(
    x_val,
    y_val,
    batch_size=32,
    shuffle=False
)
```

**Insights:**

Data augmentation was applied to the training data to simulate real-world variations such as slight rotations, shifts, zoom, and lighting changes. This helps the model generalize better and reduces overfitting by exposing it to more diverse facial patterns during training, while validation data remained unchanged to ensure unbiased evaluation.

# B. ResNet50 :-

```python
checkpointer = ModelCheckpoint(
    filepath='/content/drive/MyDrive/Colab Notebooks/gender_detection_best_resnet_model.keras',
    monitor='val_loss',
    save_best_only=True,
    mode='min',
    verbose=1
)

early_stop = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True,
    verbose=1
)

lr_reduce = ReduceLROnPlateau(
    monitor='val_loss',
    patience=3,
    factor=0.3,
    min_lr=1e-6,
    verbose=1
)
```

```python
base_model = ResNet50(
    input_shape=(224,224,3),
    include_top=False,
    weights='imagenet'
)

base_model.trainable = False
```

```python
resnet_model = models.Sequential()

resnet_model.add(base_model)

resnet_model.add(layers.GlobalAveragePooling2D())

resnet_model.add(layers.Dense(128, activation='relu'))
resnet_model.add(layers.Dropout(0.4))

resnet_model.add(layers.Dense(1, activation='sigmoid'))
```

```python
resnet_model.summary()
```

```python
resnet_model.compile(
    optimizer=Adam(learning_rate=1e-4),
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
history_resnet = resnet_model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=40,
    callbacks=[checkpointer, early_stop, lr_reduce],
    verbose=2
)
```

```python
 # Plot training metrics
acc = history_resnet.history['accuracy']
val_acc = history_resnet.history['val_accuracy']
loss = history_resnet.history['loss']
val_loss = history_resnet.history['val_loss']
epochs_range = range(len(acc))

plt.figure(figsize=(14, 6))
plt.subplot(1, 2, 1)
plt.plot(epochs_range, acc, label='Training Accuracy')
plt.plot(epochs_range, val_acc, label='Validation Accuracy')
plt.legend()
plt.title('Training and Validation Accuracy')

plt.subplot(1, 2, 2)
plt.plot(epochs_range, loss, label='Training Loss')
plt.plot(epochs_range, val_loss, label='Validation Loss')
plt.legend()
plt.title('Training and Validation Loss')
plt.show()
```

```python
resnet_model.evaluate(x_test, y_test)
```

```python
y_pred_prob = resnet_model.predict(x_test)

y_pred = (y_pred_prob > 0.5).astype("int32").flatten()

print(classification_report(
    y_test,
    y_pred,
    zero_division=0
))
```

> Using the pretrained ResNet50 architecture with transfer learning significantly improved the model’s performance for gender classification. The model achieved an accuracy of approximately 96% on the test dataset, with balanced precision, recall, and F1-scores for both classes. This indicates that the network successfully learned discriminative facial features and generalized well across the dataset. The strong performance demonstrates the effectiveness of deep residual networks and transfer learning when working with relatively small image datasets.
